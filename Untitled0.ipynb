```python
import pandas as pd

file_path = "sample_-_superstore.xlsx"
```

```python
orders = pd.read_excel(file_path, sheet_name="Orders")
people = pd.read_excel(file_path, sheet_name="People")
returns = pd.read_excel(file_path, sheet_name="Returns")

print("Orders:", orders.shape)
print("People:", people.shape)
print("Returns:", returns.shape)
```

```python
orders.isnull().sum()
```

```python
orders["Order Date"] = pd.to_datetime(
    orders["Order Date"],
    errors="coerce"
)

orders["Ship Date"] = pd.to_datetime(
    orders["Ship Date"],
    errors="coerce"
)

orders["Sales"] = pd.to_numeric(
    orders["Sales"],
    errors="coerce"
)

orders["Quantity"] = pd.to_numeric(
    orders["Quantity"],
    errors="coerce"
)

orders["Discount"] = pd.to_numeric(
    orders["Discount"],
    errors="coerce"
)

orders["Profit"] = pd.to_numeric(
    orders["Profit"],
    errors="coerce"
)
```

```python
missing = orders.isnull().sum()

print(missing[missing > 0])
```

```python
print("Exact duplicate rows:",
      orders.duplicated().sum())
```

```python
text_columns = orders.select_dtypes(include="object").columns

for col in text_columns:
    orders[col] = orders[col].str.strip()
```

```python
print("Invalid quantity:",
      (orders["Quantity"] <= 0).sum())

print("Invalid sales:",
      (orders["Sales"] <= 0).sum())

print("Invalid discount:",
      ((orders["Discount"] < 0) |
       (orders["Discount"] > 1)).sum())
```

```python
print("Order date:",
      orders["Order Date"].min(),
      "to",
      orders["Order Date"].max())

print("Ship date:",
      orders["Ship Date"].min(),
      "to",
      orders["Ship Date"].max())
```

```python
invalid_dates = orders[
    orders["Ship Date"] < orders["Order Date"]
]

print(invalid_dates)
```

```python
categorical_columns = [
    "Ship Mode",
    "Segment",
    "Country/Region",
    "Region",
    "Category",
    "Sub-Category"
]

for col in categorical_columns:
    print("\n", col)
    print(orders[col].value_counts())
```

```python
missing_orders = returns[
    ~returns["Order ID"].isin(orders["Order ID"])
]

print(missing_orders)
```

```python
cleaned_orders = orders.copy()
```

```python

# Remove whitespace from text fields
text_columns = cleaned_orders.select_dtypes(
    include="object"
).columns

for col in text_columns:
    cleaned_orders[col] = cleaned_orders[col].str.strip()

# Convert dates
cleaned_orders["Order Date"] = pd.to_datetime(
    cleaned_orders["Order Date"],
    errors="coerce"
)

cleaned_orders["Ship Date"] = pd.to_datetime(
    cleaned_orders["Ship Date"],
    errors="coerce"
)

# Convert numerical columns
numeric_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

for col in numeric_columns:
    cleaned_orders[col] = pd.to_numeric(
        cleaned_orders[col],
        errors="coerce"
    )
```

```python
cleaned_orders = cleaned_orders.drop_duplicates()
```

```python
print("Missing values:")
print(cleaned_orders.isnull().sum())

print("\nDuplicate rows:")
print(cleaned_orders.duplicated().sum())

print("\nInvalid quantities:")
print((cleaned_orders["Quantity"] <= 0).sum())

print("\nInvalid discounts:")
print(
    ((cleaned_orders["Discount"] < 0) |
     (cleaned_orders["Discount"] > 1)).sum()
)

print("\nInvalid dates:")
print(
    (cleaned_orders["Ship Date"] <
     cleaned_orders["Order Date"]).sum()
)
```

```python
change_log = pd.DataFrame({
    "Issue": [
        "Missing values",
        "Exact duplicate rows",
        "Text whitespace",
        "Invalid numeric values",
        "Invalid dates",
        "Data types"
    ],
    "Action Taken": [
        "Checked all columns; none found",
        "Checked and removed exact duplicates",
        "Trimmed leading/trailing whitespace",
        "Converted numeric columns and validated ranges",
        "Converted dates and checked ship date >= order date",
        "Standardized dates and numeric columns"
    ],
    "Result": [
        "0 missing values",
        "0 exact duplicates",
        "No problematic whitespace found",
        "No invalid values found",
        "No invalid date relationships found",
        "Validated"
    ]
})

print(change_log)
```

```python
with pd.ExcelWriter("cleaned_superstore.xlsx",
                    engine="openpyxl") as writer:

    cleaned_orders.to_excel(
        writer,
        sheet_name="Orders",
        index=False
    )

    people.to_excel(
        writer,
        sheet_name="People",
        index=False
    )

    returns.to_excel(
        writer,
        sheet_name="Returns",
        index=False
    )

    change_log.to_excel(
        writer,
        sheet_name="Change Log",
        index=False
    )
```
